```python
def course_information_agent(user_input):

    prompt = f"""
You are a Course Information Agent for Federal Urdu University,
Computer Science Department, Semester 6th B.

Your task is to provide simple and helpful information about the courses
studied in Semester 6th B.

The courses are:

1. Artificial Intelligence
2. Professional Practice
3. Compiler Construction
4. Numerical and Symbolic Computing
5. Natural Language Processing

Provide a well-structured response with clear headings:

1. Course Name
2. Course Overview
3. Main Topics Covered
4. Importance of the Course
5. Simple Study Tips

Make the response simple, short, and easy for students to understand.

User request: {user_input}

Course Information:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Semester 6th B Course Information Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = course_information_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
